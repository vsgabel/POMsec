```python
import numpy as np
import matplotlib.pyplot as plt
```

# S1

```python
t, mke, mpe = np.loadtxt("../s1/output/diagnose.out", unpack=True)
```

```python
plt.figure(figsize=(8, 6))
plt.subplot(121)
plt.title("Kinetic Energy")
plt.plot(t, mke, 'r')
plt.plot([1,1], [0, np.max(mke)*1.1], 'k--')
plt.xlabel("Time (days)")
plt.ylabel("Kinetic Energy")
plt.xlim(0, 2)
plt.ylim(0, np.max(mke)*1.1)
plt.grid()

plt.subplot(122)
plt.title("Potential Energy")
plt.plot(t, mpe, 'r')
plt.plot([1,1], [0, np.max(mpe)*1.1], 'k--')
plt.xlabel("Time (days)")
plt.ylabel("Potential Energy")
plt.xlim(0, 2)
plt.ylim(0, np.max(mpe)*1.1)
plt.grid()

plt.tight_layout()
plt.show()
```
