```python
import numpy as np
import matplotlib.pyplot as plt
from cmocean import cm
```

# S1

```python
x = np.loadtxt("../s1/output/X.DAT")
z = np.loadtxt("../s1/output/Z.DAT")
v = np.loadtxt("../s1/output/V_VEL.DAT")
bat = np.loadtxt("../s1/bat.xhd")
```

```python
plt.figure()
plt.title("Meridional Velocity - Final Output")
plt.contourf(x/1e3,z,v, cmap=cm.balance_r, levels=np.linspace(-0.5, 0.5, 41)*5e-5, extend='both')
plt.colorbar(label='Meridional Velocity (m s$^{-1}$)')
plt.fill_between(bat[:,0], -bat[:,1], -3500, color='k')
plt.xlim(0, 593)
plt.ylim(-3200,0)
plt.xlabel('Distance (km)')
plt.ylabel('Depth (m)')
plt.show()
```
